```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# 1. 데이터 정의 (엑셀 실제 데이터)
stair_speeds    = [0.178, 0.307, 0.261, 0.267, 0.224, 0.273, 0.246]
platform_speeds = [0.304, 0.800, 0.960, 1.000, 1.200]

# 2. 스타일 설정 (Times New Roman)
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams.update({'font.size': 20})

# 3. 그래프 생성
plt.figure(figsize=(7, 5))

# clip=(0, None) 옵션으로 음수 영역 발생 방지
sns.kdeplot(stair_speeds, fill=True, color='skyblue', label='Stair Speed PDF', alpha=0.4, clip=(0, None))
sns.kdeplot(platform_speeds, fill=True, color='salmon', label='Platform Speed PDF', alpha=0.4, clip=(0, None))

# 개별 데이터 포인트 표시 (Rug plot)
sns.rugplot(stair_speeds, color='blue', height=0.05)
sns.rugplot(platform_speeds, color='red', height=0.05)

# 축 설정: 물리적 한계인 0부터 시작하도록 강제
plt.xlim(0, 2.0) 
plt.xlabel('Speed (m/s)', fontsize=20)
plt.ylabel('Probability Density', fontsize=20)
plt.legend(prop={'size': 18})
plt.grid(True, linestyle='--', alpha=0.3)
plt.tight_layout()
plt.show()
```
